# 📊 Task 5: Personal Loan Acceptance Prediction

## 📌 Overview

This project focuses on predicting which customers are most likely to accept a personal loan offer. By leveraging machine learning techniques, the goal is to help financial institutions identify potential customers and improve targeted marketing strategies.

---

## 🎯 Objective

* Analyze customer data and behavior
* Build a classification model to predict loan acceptance
* Extract meaningful business insights to support decision-making

---

## 📂 Dataset

* **Bank Marketing Dataset** (UCI Machine Learning Repository)
* Contains customer demographic and financial information such as:

  * Age
  * Job
  * Marital Status
  * Income
  * Education

---

## 🛠️ Approach

### 1. Data Exploration

* Examined dataset structure using `.head()`, `.info()`, `.describe()`
* Identified key features influencing loan acceptance

### 2. Data Preprocessing

* Handled missing values
* Encoded categorical variables
* Prepared data for modeling

### 3. Exploratory Data Analysis (EDA)

* Visualized relationships between features
* Used:

  * Histograms
  * Box plots
  * Count plots

### 4. Model Building

* Implemented classification models:

  * Logistic Regression
  * Decision Tree

### 5. Model Evaluation

* Evaluated using:

  * Accuracy Score
  * Confusion Matrix

---

## 📈 Results & Insights

* Identified customer segments more likely to accept loans
* Key influencing factors include:

  * Income level
  * Education
  * Age group
* Model provides actionable insights for targeted marketing
